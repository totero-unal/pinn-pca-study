# Confirmación de PCA + media frente a POD a 100000 pasos

Diez semillas nuevas [2111, 2112, 2113, 2114, 2115, 2116, 2117, 2118, 2119, 2120], separadas de las exploraciones previas. Cien pasos uniformes y 100000 de continuación, sin reiniciar Adam. Pérdida solo PDE y condiciones IC/BC duras. Se conservan red 3×32, Adam 0.001, 4096 candidatos fijos, veinte snapshots cada cinco pasos, QR directo y reconstrucción de 128 puntos cada cien pasos. PCA + media determina el rango al 95 % de energía cuadrática; la media cuenta como dirección. POD usa el mismo rango y complemento Sobol. Es una comparación condicional, no de rangos adaptativos autónomos.

La comparación principal fijada antes de ejecutar es la diferencia pareada de RMSE terminal PCA + media − POD. El criterio de ventaja es que el extremo superior del bootstrap pareado al 95 % sea menor que cero (20000 remuestreos, semilla 6071). Medianas, validación y residuo son descriptivos; no se usan para elegir el paso terminal ni descartar semillas. No se mezclan cohortes.
